```python
import matplotlib.pyplot as plt

# Get top 10 states
top10 = mobile.sort_values(
    by='Total NFHS-6, 2023-24 (%)',
    ascending=False
).head(10)

plt.figure(figsize=(10,6))
plt.barh(top10['State / UT'], top10['Total NFHS-6, 2023-24 (%)'])
plt.gca().invert_yaxis()  # Highest value at the top

colors = ['darkblue'] + ['#A6CEE3'] * 9
plt.barh(
    top10['State / UT'],
    top10['Total NFHS-6, 2023-24 (%)'],
    color=colors
)

# Add value labels
for i, value in enumerate(top10['Total NFHS-6, 2023-24 (%)']):
    plt.text(value + 0.5, i, f'{value:.1f}%', va='center')

plt.title('Top 10 States/UTs by Women Mobile Phone Ownership (NFHS-6)', fontsize=14)
plt.xlabel('Women Owning a Mobile Phone (%)')
plt.ylabel('State / UT')
plt.xlim(0, 100)

plt.tight_layout()
plt.show()
```

```python
#overall improvement

categories = ['Mobile Ownership', 'Savings Account']

nfhs5 = [
    mobile['Total NFHS-5, 2019-21 (%)'].mean(),
    bank['Total NFHS-5, 2019-21 (%)'].mean()
]

nfhs6 = [
    mobile['Total NFHS-6, 2023-24 (%)'].mean(),
    bank['Total NFHS-6, 2023-24 (%)'].mean()
]

x = np.arange(len(categories))
width = 0.35

plt.figure(figsize=(7,5))

plt.bar(x - width/2, nfhs5, width, label='NFHS-5', color='lightgreen')
plt.bar(x + width/2, nfhs6, width, label='NFHS-6', color='darkgreen')

plt.xticks(x, categories)
plt.ylabel('Average (%)')
plt.title('Average Ownership Across States')
plt.legend()
```

```python
# Sort by improvement (largest at the top)
mobile_sorted = mobile.sort_values('Change in Total', ascending=False)

plt.figure(figsize=(10,12))

# Draw horizontal lines
for i in range(len(mobile_sorted)):
    plt.plot(
        [mobile_sorted['Total NFHS-5, 2019-21 (%)'].iloc[i],
         mobile_sorted['Total NFHS-6, 2023-24 (%)'].iloc[i]],
        [i, i],
        color='lightgrey',
        linewidth=2
    )

# NFHS-5 points
plt.scatter(
    mobile_sorted['Total NFHS-5, 2019-21 (%)'],
    range(len(mobile_sorted)),
    color='lightblue',
    s=60,
    label='NFHS-5'
)

# NFHS-6 points
plt.scatter(
    mobile_sorted['Total NFHS-6, 2023-24 (%)'],
    range(len(mobile_sorted)),
    color='mediumpurple',
    s=60,
    label='NFHS-6'
)

plt.yticks(range(len(mobile_sorted)), mobile_sorted['State / UT'])
plt.xlabel('Women Owning a Mobile Phone (%)')
plt.title('Change in Women Mobile Phone Ownership (NFHS-5 to NFHS-6)')
plt.legend()
plt.grid(axis='x', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Read data
df = pd.read_csv("Correlation Sheet - all metrics.csv")

# Set first column (Metric names) as index
df = df.set_index(df.columns[0])

# Transpose so rows = States, columns = Metrics
df = df.T

# Convert to numeric
df = df.apply(pd.to_numeric)

# Define variables
x = df["12. Women with 10 or more years of schooling (%)"]
y = df["93. Women having a mobile phone that they themselves use (%)"]

# State names (row index after transpose)
states = df.index

plt.figure(figsize=(10,8))

# Scatter plot
plt.scatter(x, y, s=60, color="lightgreen")

# Add state labels
for i, state in enumerate(states):
    plt.text(
        x.iloc[i] + 0.2,
        y.iloc[i] + 0.2,
        state,
        fontsize=8
    )

# Add regression line
m, b = np.polyfit(x, y, 1)
plt.plot(
    x,
    m*x + b,
    color="darkgreen",
    linewidth=2,
    label="Trend Line"
)

plt.xlabel("Women with 10 or More Years of Schooling (%)")
plt.ylabel("Women Owning Their Own Mobile Phone (%)")
plt.title("Women's Education and Mobile Phone Ownership Across States & UTs")

plt.grid(alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read data
df = pd.read_csv("Mobile Ownership- NFHS Survey Analysis.csv")

# Automatically detect columns
state_col = [c for c in df.columns if "State" in c][0]
nfhs5_col = [c for c in df.columns if "NFHS-5" in c][0]
nfhs6_col = [c for c in df.columns if "NFHS-6" in c][0]

# Calculate improvement
df["Improvement"] = df[nfhs6_col] - df[nfhs5_col]

# Median values (better than mean for quadrant plots)
x_mid = df[nfhs5_col].median()
y_mid = df["Improvement"].median()

plt.figure(figsize=(12,8))

plt.scatter(df[nfhs5_col], df["Improvement"], s=70)

# Label every state
for _, row in df.iterrows():
    plt.text(
        row[nfhs5_col] + 0.2,
        row["Improvement"] + 0.2,
        row[state_col],
        fontsize=8
    )

# Quadrant lines
plt.axvline(x_mid, color='red', linestyle='--')
plt.axhline(y_mid, color='red', linestyle='--')

# Labels
plt.xlabel("NFHS-5 Mobile Ownership (%)")
plt.ylabel("Improvement (NFHS-6 − NFHS-5, percentage points)")
plt.title("Women's Mobile Phone Ownership: Baseline vs Improvement")

# Quadrant names
xmin, xmax = plt.xlim()
ymin, ymax = plt.ylim()

plt.text(xmin+1, ymax-1, "Catch-up\nPerformers",
         fontsize=10, weight='bold')

plt.text(xmax-8, ymax-1, "Leaders",
         fontsize=10, weight='bold')

plt.text(xmin+1, ymin+1, "Lagging",
         fontsize=10, weight='bold')

plt.text(xmax-12, ymin+1,
         "High Baseline\nLower Improvement",
         fontsize=10, weight='bold')

plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
